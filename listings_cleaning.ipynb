```python
import re
def clean_item_age(value):

    # Missing values ko handle karo
    if pd.isna(value):
        return np.nan

    # Value ko string aur lowercase mein convert karo
    value = str(value).strip().lower()

    # Empty ya vague text ko missing banao
    if value in ["", "old", "unknown", "n/a", "na", "none", "new"]:
        return np.nan

    # Negative values invalid hain
    if re.match(r"^-\s*\d", value):
        return np.nan

    # Years ko months mein convert karo
    year_match = re.search(
        r"(\d+(?:\.\d+)?)\s*(?:years?|yrs?|yr)\b",
        value
    )

    if year_match:
        return float(year_match.group(1)) * 12

    # Months ko extract karo
    month_match = re.search(
        r"(\d+(?:\.\d+)?)\s*(?:months?|mos?|mo)\b",
        value
    )

    if month_match:
        return float(month_match.group(1))

    # Agar value simple number hai
    if re.fullmatch(r"\d+(?:\.\d+)?", value):
        return float(value)

    # Unrecognised text ko missing rakho
    return np.nan


# Step 3: Original column ka backup
df["item_age_months_original"] = df["item_age_months"]

# Step 4: Cleaning apply karo
df["item_age_months"] = df["item_age_months"].apply(
    clean_item_age
)

# Step 5: Suspiciously high values check karo
# Is dataset mein 300 months jaise values ko verify karna hai
suspicious = df["item_age_months"] > 120

print("Suspicious values:", suspicious.sum())

# Dataset ke context mein 120 months se zyada
# values ko verify kiye bina retain mat karo
df.loc[suspicious, "item_age_months"] = np.nan

# Step 6: Results check karo
print("\nCleaned values:")
print(df["item_age_months"].value_counts().sort_index().head(20))

print("\nMissing values after cleaning:")
print(df["item_age_months"].isna().sum())

print("\nData type:")
print(df["item_age_months"].dtype)

```
